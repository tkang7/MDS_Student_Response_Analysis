```python
# trendline
column_starts_with = 'Q2' 
filtered_df_block_1 = df_block_1.loc[:, df_block_1.columns.str.startswith(column_starts_with)]
filtered_df_block_1.dropna(inplace=True)

filtered_df_block_2 = df_block_2.loc[:, df_block_2.columns.str.startswith(column_starts_with)]
filtered_df_block_2.dropna(inplace=True)

filtered_df_block_3 = df_block_3.loc[:, df_block_3.columns.str.startswith(column_starts_with)]
filtered_df_block_3.dropna(inplace=True)
```

```python
df_block_1_value_count = filtered_df_block_1.apply(pd.Series.value_counts)
df_block_2_value_count = filtered_df_block_2.apply(pd.Series.value_counts)
df_block_3_value_count = filtered_df_block_3.apply(pd.Series.value_counts)

df_combined = pd.concat([df_block_1_value_count, df_block_2_value_count, df_block_3_value_count], ignore_index=False, axis=1)
df_combined
```

```python
def convert_values(df):
    satisfaction_mapping = {
        "Very dissatisfied": -2,
        "Dissatisfied": -1,
        "Neutral": 0,
        "Satisfied": 1,
        "Very satisfied": 2
    }   

    q5_mapping = {
        "Extremely Challenging": -2,
        "Frequently Overwhelming": -1,
        "Sometimes Overwhelming": 0,
        "Mostly Manageable": 1,
        "Very Manageable": 2
    }

    q6_mapping = {
        "Extremely difficult to manage": -2,
        "Frequently Overwhelming": -1,
        "Sometimes Overwhelming": 0,
        "Mostly Manageable": 1,
        "Very Manageable": 2
    }

    challenging_column = df.columns[df.columns.str.startswith("Q6")]
    satisfaction_column = df.columns.difference(challenging_column)

    df[satisfaction_column] = df[satisfaction_column].applymap(lambda x: satisfaction_mapping.get(x, x))
    df[challenging_column] = df[challenging_column].applymap(lambda x: q6_mapping.get(x, x))

    return df
    
```

```python
df_block_1_mapped = convert_values(filtered_df_block_1)
df_block_1_mapped
```
